# Temporizador / Cronómetro

Proyecto en **Python puro** (`time` + `threading`).

- **Cronómetro**: cuenta hacia arriba (presiona Enter para detener)
- **Temporizador**: cuenta hacia atrás y avisa al terminar

Ejecuta las celdas en orden.

```python
import time
import threading

def formato_tiempo(segundos: float) -> str:
    """Convierte segundos a HH:MM:SS."""
    segundos = int(segundos)
    h = segundos // 3600
    m = (segundos % 3600) // 60
    s = segundos % 60
    return f"{h:02d}:{m:02d}:{s:02d}"

def cronometro():
    """Cronómetro que cuenta hacia arriba. Presiona Enter para detener."""
    print("Cronómetro iniciado. Presiona Enter para detener.\n")
    inicio = time.time()
    detenido = threading.Event()

    def esperar_enter():
        input()
        detenido.set()

    hilo = threading.Thread(target=esperar_enter, daemon=True)
    hilo.start()

    try:
        while not detenido.is_set():
            transcurrido = time.time() - inicio
            print(f"\rTiempo: {formato_tiempo(transcurrido)}", end="", flush=True)
            time.sleep(0.1)
    except KeyboardInterrupt:
        pass

    print(f"\n\nTiempo final: {formato_tiempo(time.time() - inicio)}")

def temporizador(segundos: int):
    """Temporizador que cuenta hacia atrás."""
    print(f"Temporizador de {formato_tiempo(segundos)} iniciado.\n")
    inicio = time.time()
    fin = inicio + segundos

    try:
        while time.time() < fin:
            restante = fin - time.time()
            print(f"\rRestante: {formato_tiempo(restante)}", end="", flush=True)
            time.sleep(0.1)
    except KeyboardInterrupt:
        print("\n\nTemporizador cancelado.")
        return

    print("\n\n¡Tiempo terminado!")
    print("\a", end="", flush=True)

print("Funciones cargadas: cronometro() y temporizador()")
```

## Temporizador corto de prueba (10 segundos)

```python
temporizador(10)
```

## Cronómetro
Ejecuta y presiona Enter cuando quieras detenerlo.

```python
cronometro()
```

## Modo interactivo

```python
print("1. Cronómetro  |  2. Temporizador\n")
opcion = input("Opción: ").strip()

if opcion == "1":
    cronometro()
elif opcion == "2":
    minutos = int(input("Minutos (defecto 0): ") or 0)
    segundos = int(input("Segundos (defecto 30): ") or 30)
    total = minutos * 60 + segundos
    if total > 0:
        temporizador(total)
    else:
        print("El tiempo debe ser mayor a 0.")
else:
    print("Opción no válida.")
```
