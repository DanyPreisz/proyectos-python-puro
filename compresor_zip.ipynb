# Compresor / Descompresor ZIP

Proyecto en **Python puro** usando `zipfile` (biblioteca estándar).

Funciones:
- Comprimir archivos o carpetas
- Descomprimir ZIP
- Listar contenido de un ZIP

Ejecuta las celdas en orden.

```python
import zipfile
from pathlib import Path

def comprimir(archivos: list, salida: str):
    """Comprime uno o varios archivos/carpetas en un ZIP."""
    salida_path = Path(salida)
    if not salida_path.suffix:
        salida_path = salida_path.with_suffix(".zip")

    with zipfile.ZipFile(salida_path, "w", zipfile.ZIP_DEFLATED) as zf:
        for item in archivos:
            path = Path(item)
            if not path.exists():
                print(f"Aviso: no existe {item}")
                continue

            if path.is_file():
                zf.write(path, path.name)
                print(f"  + {path.name}")
            elif path.is_dir():
                for archivo in path.rglob("*"):
                    if archivo.is_file():
                        arcname = archivo.relative_to(path.parent)
                        zf.write(archivo, arcname)
                        print(f"  + {arcname}")

    print(f"\nCreado: {salida_path}")
    return str(salida_path)

def descomprimir(archivo_zip: str, destino: str = "."):
    """Descomprime un archivo ZIP."""
    zip_path = Path(archivo_zip)
    if not zip_path.exists():
        raise FileNotFoundError(f"No se encontró: {archivo_zip}")

    destino_path = Path(destino)
    destino_path.mkdir(parents=True, exist_ok=True)

    with zipfile.ZipFile(zip_path, "r") as zf:
        zf.extractall(destino_path)
        print(f"Extraído en: {destino_path}")
        print("Archivos:")
        for nombre in zf.namelist():
            print(f"  - {nombre}")

    return str(destino_path)

def listar_contenido(archivo_zip: str):
    """Lista el contenido de un ZIP sin extraerlo."""
    zip_path = Path(archivo_zip)
    if not zip_path.exists():
        raise FileNotFoundError(f"No se encontró: {archivo_zip}")

    with zipfile.ZipFile(zip_path, "r") as zf:
        print(f"Contenido de {zip_path.name}:\n")
        for info in zf.infolist():
            print(f"  {info.filename}  ({info.file_size} bytes)")

print("Funciones cargadas: comprimir(), descomprimir() y listar_contenido()")
```

## Ejemplo rápido
Crea un archivo de prueba, lo comprime y lista el contenido.

```python
# Crear un archivo de prueba
Path("ejemplo.txt").write_text("Este es un archivo de prueba para comprimir.\nLínea 2.")

# Comprimir
comprimir(["ejemplo.txt"], "ejemplo.zip")

print()
# Listar contenido
listar_contenido("ejemplo.zip")
```

## Descomprimir
Ejecuta esta celda para extraer el ZIP de ejemplo.

```python
descomprimir("ejemplo.zip", "extraido")
```

## Modo interactivo

```python
print("1. Comprimir  |  2. Descomprimir  |  3. Listar\n")
opcion = input("Opción: ").strip()

if opcion == "1":
    archivos = input("Archivos (separados por espacio): ").split()
    salida = input("Nombre del ZIP: ").strip() or "archivo.zip"
    comprimir(archivos, salida)
elif opcion == "2":
    archivo = input("Archivo ZIP: ").strip()
    destino = input("Destino (Enter = actual): ").strip() or "."
    descomprimir(archivo, destino)
elif opcion == "3":
    archivo = input("Archivo ZIP: ").strip()
    listar_contenido(archivo)
else:
    print("Opción no válida.")
```
