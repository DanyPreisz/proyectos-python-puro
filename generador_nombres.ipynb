# Generador de Nombres

Proyecto en **Python puro** (`random`).

Genera:
- Nombres de usuario (`adjetivo_sustantivo123`)
- Nombres de proyectos (`proyecto-adjetivo-sustantivo`)
- Nombres completamente aleatorios

Ejecuta las celdas en orden.

```python
import random
import string

ADJETIVOS = [
    "rapido", "brillante", "silencioso", "poderoso", "creativo",
    "oscuro", "luminoso", "agil", "sabio", "valiente",
    "misterioso", "feliz", "audaz", "sereno", "feroz",
]

SUSTANTIVOS = [
    "lobo", "aguila", "dragon", "tigre", "leon",
    "codigo", "nube", "estrella", "rio", "montana",
    "fuego", "viento", "oceano", "rayo", "sombra",
]

PROYECTOS_PREFIJOS = [
    "proyecto", "app", "sistema", "herramienta", "plataforma",
    "servicio", "modulo", "libreria", "bot", "api",
]

def generar_usuario(cantidad: int = 1, con_numero: bool = True, longitud_numero: int = 2) -> list:
    resultados = []
    for _ in range(cantidad):
        adj = random.choice(ADJETIVOS)
        sust = random.choice(SUSTANTIVOS)
        nombre = f"{adj}_{sust}"
        if con_numero:
            numero = "".join(random.choices(string.digits, k=longitud_numero))
            nombre += numero
        resultados.append(nombre)
    return resultados

def generar_proyecto(cantidad: int = 1, con_numero: bool = False) -> list:
    resultados = []
    for _ in range(cantidad):
        prefijo = random.choice(PROYECTOS_PREFIJOS)
        adj = random.choice(ADJETIVOS)
        sust = random.choice(SUSTANTIVOS)
        nombre = f"{prefijo}-{adj}-{sust}"
        if con_numero:
            nombre += f"-{random.randint(1, 99)}"
        resultados.append(nombre)
    return resultados

def generar_aleatorio(cantidad: int = 1, longitud: int = 8) -> list:
    caracteres = string.ascii_lowercase + string.digits
    return [
        "".join(random.choices(caracteres, k=longitud))
        for _ in range(cantidad)
    ]

print("Funciones cargadas: generar_usuario(), generar_proyecto() y generar_aleatorio()")
```

## Ejemplos rápidos

```python
print("=== Usuarios ===")
for n in generar_usuario(5):
    print(f"  {n}")

print("\n=== Proyectos ===")
for n in generar_proyecto(5):
    print(f"  {n}")

print("\n=== Aleatorios ===")
for n in generar_aleatorio(5, 10):
    print(f"  {n}")
```

## Modo interactivo

```python
print("1. Usuarios  |  2. Proyectos  |  3. Aleatorios\n")
opcion = input("Opción: ").strip()
cantidad = int(input("¿Cuántos? (defecto 5): ") or 5)

if opcion == "1":
    nombres = generar_usuario(cantidad)
elif opcion == "2":
    nombres = generar_proyecto(cantidad)
elif opcion == "3":
    longitud = int(input("Longitud (defecto 8): ") or 8)
    nombres = generar_aleatorio(cantidad, longitud)
else:
    nombres = []
    print("Opción no válida.")

if nombres:
    print("\n--- Nombres generados ---\n")
    for i, n in enumerate(nombres, 1):
        print(f"{i}. {n}")
```
