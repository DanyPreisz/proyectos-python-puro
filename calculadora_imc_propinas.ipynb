# Calculadora de IMC y Propinas

Proyecto en **Python puro**.

- Calcula el Índice de Masa Corporal (IMC) y su clasificación
- Calcula propina, total y monto por persona

Ejecuta las celdas en orden.

```python
def calcular_imc(peso: float, altura: float) -> dict:
    """Calcula el IMC y su clasificación."""
    if peso <= 0 or altura <= 0:
        raise ValueError("Peso y altura deben ser mayores a 0.")

    imc = peso / (altura ** 2)

    if imc < 18.5:
        clasificacion = "Bajo peso"
    elif imc < 25:
        clasificacion = "Peso normal"
    elif imc < 30:
        clasificacion = "Sobrepeso"
    else:
        clasificacion = "Obesidad"

    return {"imc": round(imc, 2), "clasificacion": clasificacion}

def calcular_propina(total: float, porcentaje: float, personas: int = 1) -> dict:
    """Calcula propina y total por persona."""
    if total < 0 or porcentaje < 0 or personas < 1:
        raise ValueError("Valores inválidos.")

    propina = total * (porcentaje / 100)
    total_con_propina = total + propina
    por_persona = total_con_propina / personas

    return {
        "propina": round(propina, 2),
        "total": round(total_con_propina, 2),
        "por_persona": round(por_persona, 2),
    }

print("Funciones cargadas: calcular_imc() y calcular_propina()")
```

## Ejemplos rápidos

```python
# IMC
print(calcular_imc(70, 1.75))
print(calcular_imc(55, 1.60))

# Propina
print(calcular_propina(1500, 10, 3))
print(calcular_propina(850, 15, 2))
```

## Modo interactivo

```python
print("1. IMC  |  2. Propina\n")
opcion = input("Opción: ").strip()

if opcion == "1":
    peso = float(input("Peso (kg): "))
    altura = float(input("Altura (m): "))
    r = calcular_imc(peso, altura)
    print(f"\nIMC: {r['imc']}")
    print(f"Clasificación: {r['clasificacion']}")
elif opcion == "2":
    total = float(input("Total de la cuenta: "))
    porcentaje = float(input("Porcentaje de propina (defecto 10): ") or 10)
    personas = int(input("Número de personas (defecto 1): ") or 1)
    r = calcular_propina(total, porcentaje, personas)
    print(f"\nPropina: {r['propina']}")
    print(f"Total con propina: {r['total']}")
    print(f"Por persona: {r['por_persona']}")
else:
    print("Opción no válida.")
```
