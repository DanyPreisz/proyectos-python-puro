# Organizador de Carpetas por Extensión

Proyecto en **Python puro** (`pathlib`).

Mueve los archivos de una carpeta a subcarpetas según su extensión:
- `foto.jpg` → `jpg/foto.jpg`
- `documento.pdf` → `pdf/documento.pdf`
- `script.py` → `py/script.py`

Incluye **modo simulación** para ver los cambios antes de aplicarlos.

Ejecuta las celdas en orden.

```python
from pathlib import Path

def organizar_por_extension(
    carpeta: str,
    simulacion: bool = True,
) -> list:
    """Mueve archivos a subcarpetas según su extensión."""
    carpeta_path = Path(carpeta)
    if not carpeta_path.exists() or not carpeta_path.is_dir():
        raise ValueError(f"Carpeta no válida: {carpeta}")

    cambios = []

    for archivo in carpeta_path.iterdir():
        if not archivo.is_file():
            continue

        extension = archivo.suffix.lower().lstrip(".") or "sin_extension"
        destino_carpeta = carpeta_path / extension
        destino = destino_carpeta / archivo.name
        cambios.append((archivo, destino))

    if not simulacion:
        for original, destino in cambios:
            if original != destino:
                destino.parent.mkdir(exist_ok=True)
                original.rename(destino)

    return cambios

def mostrar_cambios(cambios: list, simulacion: bool = True):
    if not cambios:
        print("No hay archivos para organizar.")
        return

    modo = "SIMULACIÓN" if simulacion else "EJECUTADO"
    print(f"\n[{modo}] {len(cambios)} archivo(s):\n")
    for original, destino in cambios:
        print(f"  {original.name}  →  {destino.parent.name}/{destino.name}")
    print()

print("Funciones cargadas: organizar_por_extension() y mostrar_cambios()")
```

## Ejemplo (solo simulación)
Cambia la carpeta si es necesario. En Colab suele ser `/content`.

```python
# Simulación segura (no mueve nada)
cambios = organizar_por_extension(carpeta=".", simulacion=True)
mostrar_cambios(cambios, simulacion=True)

# Para aplicar de verdad:
# cambios = organizar_por_extension(carpeta=".", simulacion=False)
# mostrar_cambios(cambios, simulacion=False)
```

## Modo interactivo
Primero muestra la simulación y luego pregunta si quieres aplicar.

```python
print("=== Organizador de Carpetas ===\n")
carpeta = input("Carpeta a organizar (Enter = actual): ").strip() or "."

cambios = organizar_por_extension(carpeta, simulacion=True)
mostrar_cambios(cambios, simulacion=True)

if cambios:
    confirmar = input("¿Aplicar cambios? (s/n): ").lower()
    if confirmar == "s":
        organizar_por_extension(carpeta, simulacion=False)
        print("Archivos organizados.")
    else:
        print("Cancelado.")
```
