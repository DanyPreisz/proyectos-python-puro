# Calculadora en Python Puro\n\nProyecto de **calculadora interactiva** usando solo la biblioteca estándar de Python.\nSin frameworks ni dependencias externas.\n\n## Características\n- Modo interactivo\n- Evaluación de expresiones: `+ - * / **` (potencia), `sqrt()`, porcentaje\n- Manejo de errores\n- Lista para Google Colab\n\n## Instrucciones\n1. Ejecuta las celdas en orden.\n2. En la celda interactiva, escribe expresiones y presiona Enter.\n3. Escribe `salir` para terminar.\n

```python
import math\nimport sys\n\ndef calcular(expresion: str) -> float:\n    \"\"\"Evalúa una expresión matemática de forma segura.\"\"\"\n    # Normalizar funciones y porcentaje\n    expresion = expresion.replace(\"sqrt\", \"math.sqrt\")\n    expresion = expresion.replace(\"%\", \"/100*\")  # 25% -> 25/100*\n\n    allowed = {\n        \"math\": math,\n        \"__builtins__\": {},\n    }\n\n    try:\n        resultado = eval(expresion, allowed)\n        return float(resultado)\n    except Exception as e:\n        raise ValueError(f\"Error en la expresión: {e}\")\n\ndef modo_interactivo():\n    print(\"=== Calculadora Python Puro ===\")\n    print(\"Operaciones: + - * / ** sqrt() %\")\n    print(\"Ejemplos: 5 + 3, sqrt(16), 20%, 2 ** 8\")\n    print(\"Escribe 'salir' para terminar.\\n\")\n\n    while True:\n        try:\n            entrada = input(\">>> \").strip()\n            if entrada.lower() in (\"salir\", \"exit\", \"q\"):\n                print(\"Hasta luego!\")\n                break\n            if not entrada:\n                continue\n\n            resultado = calcular(entrada)\n            print(f\"= {resultado}\\n\")\n        except KeyboardInterrupt:\n            print(\"\\nHasta luego!\")\n            break\n        except Exception as e:\n            print(f\"Error: {e}\\n\")\n\nprint(\"Funciones cargadas: calcular() y modo_interactivo()\")
```

## Ejemplos rápidos

```python
print(calcular(\"10 + 5\"))\nprint(calcular(\"sqrt(81)\"))\nprint(calcular(\"2 ** 10\"))\nprint(calcular(\"25%\"))\nprint(calcular(\"(5 + 3) * 2\"))
```
